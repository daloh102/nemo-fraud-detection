## 2. Schritt 4: Qualitätskontrolle & Daten-Inspektion

Wir untersuchen die ersten erzeugten Einträge in `data/raw/call_transcripts_raw.jsonl` sowie die erzeugten Kontaminationsarten.

```python
raw_path = "../data/raw/call_transcripts_raw.jsonl"
benchmark_path = "../data/evaluation/benchmark_dataset.jsonl"

# 1. Zeige das allererste Rohdaten-Transkript
with open(raw_path, "r", encoding="utf-8") as f:
    first_record = json.loads(f.readline())

print(f"🆔 Call ID: {first_record['call_id']}")
print("=" * 60)
print(first_record['text'])
print("=" * 60)
```

```python
# 2. Übersicht über injizierten Noise/Kontamination im Test-Set
with open(raw_path, "r", encoding="utf-8") as f:
    records = [json.loads(line) for line in f]

print(f"📊 Gesamtanzahl erzeugter Einträge (inkl. Duplikate): {len(records)}\n")

print("🔍 Stichprobe verschiedener Einträge:")
for i, rec in enumerate(records[:5]):
    preview = rec['text'].replace('\n', ' | ')[:90]
    print(f"[{i+1}] ID: {rec['call_id']:<15} Text: {preview}...")
```
